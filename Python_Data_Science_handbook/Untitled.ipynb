### Multi dimensional arrays

Just a list = a 1D array

A llist of lists = 2D array

List of lists of lists = 3D array .... and so on

```python
a = np.array([[1,2,3,4,5],
            [10,11,12,13,14]])
print(a)
print("Shape of a is:", a.shape)
print("Size of a is:",a.size)
print("# of dimensions of a:", a.ndim)
```
