At the implementation level, the array essentially contains a single pointer to one contiguous block of data. The Python list, on the other hand, contains a pointer to a block of pointers, each of which in turn points to a full Python object like the Python integer we saw earlier. Again, the advantage of the list is flexibility: because each list element is a full structure containing both data and type information, the list can be filled with data of any desired type. Fixed-type NumPy-style arrays lack this flexibility, but are much more efficient for storing and manipulating data.

```python
import array
L = list(range(10))
A = array.array("i",L)      #Here, 'i' is a type code indicating the contents are integers.
A
```

```python
import numpy as np
```

```python
#we can use np.array to create arrays from Python lists
np.array([1,4,2,5,3])
```

NumPy arrays can only contain data of the same type. If the types do not match, NumPy will upcast them according to its type promotion rules; here, integers are upcast to floating point

```python
np.array([3.14,1,4,3,2])
```

If we want to explicitly set the data type of the resulting array, we can use the dtype keyword

```python
np.array([1,2,3,4], dtype = np.float32)
```

```python
y = np.array([5,6,78.78,8],dtype = float)
print(y)
y
```

```python
#Python elements of lists
for i in [2,4,6]:
    print(list(range(i,i+3)))
```

```python
#Python lists of lists
L = []
for i in [2,4,6]:
    L.append(list(range(i,i+3)))
print(L)
```

```python
#numpy arrays
rows = []
for i in [2,4,6]:
    rows.append(range(i,i+3))
print(rows)
np.array(rows)
```

```python
#numpy arrays - condensed version using list comprehensions
x = np.array([range(i,i+3) for i in [2,4,6]])
x
```

```python
type(x)
```

```python
x.shape #3x3 matrix
```

### Creating arrays from scratch

```python
# Create a length-10 integer array filled with 0s
x = np.zeros(10,dtype = int)
x
```

```python
# Create a 3x5 floating-point array filled with 0s
np.zeros((2,3),dtype = int)
```

```python
# Create a 3x5 floating-point array filled with 1s
y = np.ones((3,5),dtype = float)
print(y)
```
