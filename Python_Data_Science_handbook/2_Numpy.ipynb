### Data types in python

```python
/* C code */
int result = 0;
for(int i=0; i<100; i++){
    result += i;
}
```

```python
# Python code
result = 0
for i in range(100):
    result += i
```

Unlike in C, in Python there is no need to name the data type of the variable. 

 a C integer is essentially a label for a position in memory whose bytes encode an integer value. A Python integer is a pointer to a position in memory containing all the Python object information, including the bytes that contain the integer value. 

```python
L1 = list(range(10))
L2 = [str(x) for x in L1]

print(L1)
print(L2)
```

```python
L3 = [True, 120, 3.14,"alpha" ]
[type(c) for c in L3]
```

: to allow these flexible types, each item in the list must contain its own type, reference count, and other information. That is, each item is a complete Python object.

### Numpy array attributes



DEfine random raays of 1, 2 and 3 dimensions. 

```python
import numpy as np

rng = np.random.default_rng(seed = 1000)

x1 = rng.integers(10,size = 6)      # 1D array
x2 = rng.integers(10, size = (3,4)) # 2D array
x3 = rng.integers(10, size = (3,4,5)) #3D array
```

Each array has attributes: `ndim` (no. pf dimensions); `shape` (size of each dimension); `size` (total size of the array) and `dtype` (the type of each array element)

```python
print(x1)
print("__________________________________")
print(x2)
print("__________________________________")
print(x3)
```
