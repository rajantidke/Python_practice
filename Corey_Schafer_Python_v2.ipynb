#### Significance of "and" vs "or"

```python
user = "Admin"
logged_in = True

if user == "Admin" and logged_in:
    print("Admin page")
else:
    print("Wrong credentials")
```

```python
user = "Admin"
logged_in = False

if user == "Admin" and logged_in:
    print("Admin page")
else:
    print("Wrong credentials")
```

```python
user = "Admin"
logged_in = False

if user == "Admin" or logged_in:
    print("Atleast one admin credentials is right")
else:
    print("Admin credentials are both wrong")
```

```python
user = "Admin"
logged_in = False

if not logged_in:
    print("Please log in.")
else:
    print("Welcome to admin page")
```
