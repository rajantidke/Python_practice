### List Comprehensions

```python
mystring = "hello"

mylist = []

for letter in mystring:
    mylist.append(letter)

print(mylist)
```

```python
mystring = "Alpha"
mylist = [letter for letter in mystring]
print(mylist)
```

```python
mylist = [num for num in range(1,11)]
print(mylist)
```

```python
my_squared_list = [num**2 for num in range(1,11)]
print(mylist)
print(my_squared_list)
```

```python
my_even_list = [num for num in range(1,11) if num%2 == 0 ]
print(mylist)
print(my_squared_list)
print(my_even_list)
```

```python
my_even_squared_list = [num**2 for num in range(1,11) if num%2 == 0]
print(mylist)
print(my_squared_list)
print(my_even_list)
print(my_even_squared_list)
```

```python
celcius = [0,10,20,30,100]

fahrenheit = [(9/5*temp + 32) for temp in celcius]
print(fahrenheit)
```

```python
#Nested loops
mylist = []

for x in [2,4,6]:
    for y in [100,200,300]:
        mylist.append(x*y)
print(mylist)
```

```python
#Nested loops using list comprehensions
mylist1 = [x*y for x in [2,4,6] for y in [100,200,300]]
print(mylist1)
```

```python
# Go through the string below and if the length of a word is even print "even!"
st = 'Print every word in this sentence that has an even number of letters'

for word in st.split():
    if len(word)%2 == 0:
        print(word,"even!")
```

```python
# Write a program that prints the integers from 1 to 100. 
# But for multiples of three print "Fizz" instead of the number, 
# and for the multiples of five print "Buzz". 
# For numbers which are multiples of both three and five print "FizzBuzz".
```

```python
for num in range(1,101):
    if (num%3 == 0) and (num%5 == 0):
        print("FizzBuzz")
    elif num%3 == 0:
        print("Fizz")
    elif num%5 == 0:
        print("Buzz")
    else:
        print(num)
```
