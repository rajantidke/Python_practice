```python
#combining booleans, logics, data types, if-else, loops and functions together

# Days in the corresponding month for index 1 to 12. Index zero is zero
month_days = [0,31,28,31,30,31,30,31,31,30,31,30,31]

def is_leap(year):
    """Returns True for Leap years and false for non leap years"""
    return year % 4 == 0 and ( (year % 100 != 0) or (year % 400 == 0))

def days_in_month(year,month):
    """Return no. of days in that specific months of that specific year"""
    if not 1<= month <=12:
        return "Invalid Month"
    if month == 2 and is_leap(year):
        return 29

    return month_days[month]
```

```python
print(is_leap(2017))
```

```python
print(is_leap(2016))
```

```python
print(is_leap(2016))
```

```python
print(days_in_month(2017,2))
```

```python
print(days_in_month(2016,2))
```

```python
print(days_in_month(2017,3))
```

```python
print(days_in_month(2016,3))
```
